## Ejercicio 1: Variables, tipos y f-strings

**Referencia:** Básico 1 (línea de pedido con IVA)

**Contexto:** ahora no vendes, compras. El departamento de compras de la empresa pide material a un proveedor y necesita saber cuánto va a pagar. Un material de oficina tiene un IVA reducido del 10%. El cálculo es el mismo que en la línea de pedido de venta, pero con otros datos.

**Enunciado:**
1. Crea las variables `articulo = "Ratón"`, `precio = 12.5`, `unidades = 4` y `tipo_iva = 10`.
2. Calcula `base` (precio por unidades), `cuota_iva` (base por tipo de IVA entre 100) y `total` (base más cuota).
3. Muestra el resultado con f-strings y dos decimales, como en la salida esperada.
4. Imprime el tipo de cada variable con `type()`.

**Recuerda:** en Python no declaras el tipo de las variables ni usas `;`. Los f-strings sustituyen a `String.format` y a la concatenación con `+`. Seguimos con `float` por simplicidad, y más adelante pasaremos a `Decimal`.

**Salida esperada:**
```text
Ratón x 4
Base: 50.00€
IVA (10%): 5.00€
Total: 55.00€
<class 'str'> <class 'float'> <class 'int'>
```

````python
articulo = "Raton"
precio = 12.5
unidades = 4
tipo_iva = 10


base = precio * unidades
cuota_iva = (base * tipo_iva) / 100
total = base + cuota_iva
print(f"Base: {base}€")
print(f"IVA (10%): {cuota_iva:.2f}€")
print(f"Total: {total:.2f}€")
print(type(base), type(cuota_iva), type(total))
````

## Ejercicio 2: Condicionales y gastos de envío

**Referencia:** Básico 2 (descuento por volumen)

**Contexto:** una tienda online cobra el envío según el importe del pedido. Cuanto más compra el cliente, menos paga de envío. Es otra regla de negocio con tramos, igual que el descuento por volumen. Tienes que decidir qué rama del `if` se ejecuta según el valor.

**Enunciado:** los gastos de envío se calculan así:

- Pedido de menos de 50 €: envío de 5.95 €
- Pedido de 50 € a menos de 150 €: envío de 2.95 €
- Pedido de 150 € o más: envío gratis

Calcula `envio` y `total` (importe más envío) para tres pedidos: 30, 100 y 200. Usa un bucle `for importe in [30, 100, 200]:` para no repetir código. Presta atención a los límites: 50 entra en el segundo tramo y 150 en el tercero.

**Recuerda:** el bloque se marca con `:` y la indentación. `else if` se escribe `elif`. No hay llaves ni paréntesis obligatorios en la condición.

**Salida esperada:**
```text
Pedido 30.00€ -> envío 5.95€ -> total 35.95€
Pedido 100.00€ -> envío 2.95€ -> total 102.95€
Pedido 200.00€ -> envío 0.00€ -> total 200.00€
```

````python
for importe in [30, 100, 200]:
    if importe < 50:
        envio = 5.95
    elif 50 <= importe < 150:
        envio = 2.95
    else:
        envio = 0
    total = importe + envio
    print(f"Importe {importe:.2f}€ -> envío {envio:.2f}€ -> total {total:.2f}€") 
````

## Ejercicio 3: Bucles `for` y `while`

**Referencia:** Básico 3 (tarifa por unidades y stock)

**Contexto:** una empresa permite pagar una factura a plazos. Con un bucle `for` generas el calendario de cuotas. Además, el departamento financiero quiere saber cuántos gastos puede cubrir con el dinero de caja. Eso se resuelve con un `while`, que se repite mientras haya saldo suficiente.

**Enunciado:**

### Parte A: `for`
Una factura de 240 € se paga en 4 cuotas iguales. Calcula el valor de la cuota. Con `range(1, 5)`, muestra para cada cuota cuánto se lleva pagado y cuánto queda pendiente.

### Parte B: `while`
La caja tiene 100 € y cada gasto cuesta 30 €. Mientras la caja pueda cubrir un gasto completo (30 o más), registra el gasto y réstalo. Al final muestra cuántos gastos se pagaron y cuánto dinero queda.

**Recuerda:** `range(1, 5)` va de 1 a 4, porque el final no se incluye. Se escribe `caja -= 30`, no `caja--`. Si olvidas actualizar la variable del `while`, el bucle será infinito.

**Salida esperada:**
```text
Cuota 1: pagado 60.00€, pendiente 180.00€
Cuota 2: pagado 120.00€, pendiente 120.00€
Cuota 3: pagado 180.00€, pendiente 60.00€
Cuota 4: pagado 240.00€, pendiente 0.00€
Gastos pagados: 3, caja restante: 10
```

````python
#A
total_pagado = 240
numero_cuotas = 4
valor_cuota = total_pagado / numero_cuotas

for cuota in range(1, 5):
    pagado = cuota * valor_cuota
    pendiente = total_pagado - pagado
    print(f"Cuota {cuota}: pagado {pagado:.2f}€, pendiente {pendiente:.2f}€")
    
#B
caja = 100
gasto_caja = 30
gastos_pagados = 0

while caja >= gasto_caja:
    caja -= gasto_caja
    gastos_pagados += 1
print(f"Gastos pagados: {gastos_pagados}, caja restante: {caja} €")
````

## Ejercicio 4: Listas y operaciones básicas

**Referencia:** Básico 4 (ventas diarias)

**Contexto:** contabilidad lleva un registro de los gastos de la semana. Antes de cerrar el informe, necesita saber cuánto se ha gastado, cuál fue el gasto más pequeño y cuál es la media. Como en las ventas, los datos están en una lista y hay que resumirlos. Aquí buscamos el mínimo, no el máximo.

**Enunciado:** parte de `gastos = [60.0, 25.5, 110.0, 15.0, 80.5]`.

1. Muestra cuántos gastos hay (`len`), el total (`sum`), el menor (`min`) y la media.
2. Añade un nuevo gasto de `40.5` con `append` y recalcula el total y la media.
3. Crea una lista con los gastos superiores a 50 €. Hazlo primero con un bucle `for` y `if`. Si te animas, hazlo después con una list comprehension.

**Recuerda:** una lista de Python es parecida a un `ArrayList`. `append` equivale a `add`, y `gastos[-1]` es el último elemento. La media se calcula dividiendo la suma entre `len`.

**Salida esperada:**
```text
Nº gastos: 5
Total: 291.00€
Menor: 15.00€
Media: 58.20€
Tras añadir -> Total: 331.50€, Media: 55.25€
Superiores a 50€: [60.0, 110.0, 80.5]
```

````python
gastos = [60.0, 25.5, 110.0, 15.0, 80.5]

numero_gastos = len(gastos)
total_gastos = sum(gastos)
menor_gasto = min(gastos)
media_gastos = total_gastos / numero_gastos

print(f"Nº Gastos: {numero_gastos}")
print(f"Total: {total_gastos}€")
print(f"Menor: {menor_gasto}€")
print(f"Media: {media_gastos}€")

gastos.append(40.5)
total_gastos = sum(gastos)
print(f"Total tras añadir: {total_gastos}€")

gastos_mayores = []
for gasto in gastos:
    if gasto > 50:
        gastos_mayores.append(gasto)
print(f"Superiores a 50€: {gastos_mayores}")
````

## Ejercicio 5: Funciones reutilizables

**Referencia:** Básico 5 (IVA y email)

**Contexto:** la empresa quiere centralizar dos cálculos que se repiten: aplicar descuentos y calcular el total con envío. También quiere comprobar que un CIF de proveedor tiene el formato esperado. Si cada departamento programa su propia fórmula, aparecerán diferencias. Con funciones, la regla vive en un solo sitio.

**Enunciado:** crea estas tres funciones y pruébalas:

- `aplicar_descuento(importe, porcentaje=10)`: devuelve el importe tras restar el porcentaje, redondeado a 2 decimales con `round`. El porcentaje tiene valor por defecto de 10.
- `total_final(importe, porcentaje=10, envio=4.95)`: devuelve el importe con descuento más el envío, redondeado a 2 decimales. Debe reutilizar `aplicar_descuento`, sin repetir la fórmula.
- `es_cif_valido(cif)`: devuelve `True` si el texto tiene exactamente 9 caracteres. Si no, devuelve `False`.

**Pruebas:**
- `aplicar_descuento(200)`
- `aplicar_descuento(80, 25)`
- `total_final(200)`
- `es_cif_valido("B12345678")`
- `es_cif_valido("B1234")`
- `es_cif_valido("B123456789")`

**Recuerda:** se declaran con `def` y no necesitan estar dentro de una clase. Los parámetros con valor por defecto sustituyen a la sobrecarga de Java. Usa `len()` para medir el texto.

**Salida esperada:**
```text
180.0
60.0
184.95
True
False
False
```

````python
def aplicar_descuento(importe: float, porcentaje: float = 10) -> float:
    descuento = importe * (porcentaje / 100)
    return round(importe - descuento, 2)

def total_final(importe: float, porcentaje: float = 10, envio: float = 4.95) -> float:
    return round(aplicar_descuento(importe, porcentaje) + envio, 2)

def es_cif_valido(cif: str) -> bool:
    return len(cif) == 9

print(aplicar_descuento(200))
print(aplicar_descuento(80, 25))
print(total_final(200))
print(es_cif_valido("B12345678"))
print(es_cif_valido("B1234"))
print(es_cif_valido("B123456789"))
````
